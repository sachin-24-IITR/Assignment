```python
Q4. Why self is used in OOPs?
```

```python
In Python, self refers to the current instance of the class. It is used inside class methods to access or modify attributes and methods of the object that is calling the method. It differentiates between instance variables and local variables.
```

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand   # instance variable
        self.color = color

    def display(self):
        print(self.brand,self.color)

# Creating object
c1 = Car("Toyota", "Red")
c1.display()
```

```python
self.brand and self.color refer to the attributes of the specific object (c1).
```

```python
Q5. What is inheritance? Give an example for each type of inheritance.
```

```python
Inheritance is an important concept in Object-Oriented Programming (OOP) that allows a class (child class) to acquire the properties and behaviors (methods) of another class (parent class).

It promotes code reusability and allows new functionalities to be added with minimal changes to existing code.
```

```python
# Single inheritance
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog (Animal):
    def bark(self):
        print("Dog barks")

dog = Dog()
dog.speak()
dog.bark()
```

```python
# Multiple inheritance

class A:
  def method_A(self):
    print("Method A")

class B:
  def method_B(self):
    print("Method B")

class C(A, B):
  def method_C(self):
    print("Method C")
    print ("Multiple inheritance")
C1 = C()
C1.method_A()
C1.method_B()
C1.method_C()
```

```python
 # Multilevel inheritance
 class A:
  def method_A(self):
    print("Method A")

class B(A):
  def method_B(self):
    print("Method B")

class C(B):
  def method_C(self):
    print("Method C")
    print ("Multilevel inheritance")
C1 = C()
C1.method_A()
C1.method_B()
C1.method_C()
```

```python
# Hierarchial Inheritance
class vehicle:
  def engine(self):
    print ("This is a vehicle")
class car(vehicle):
  def car_engine(self):
    print ("This is a car")
class bike(vehicle):
  def bike_engine(self):
    print ("This is a bike")
    print ("Hierarchial inheritance")

c= car()
b= bike()
c.engine()
c.car_engine()
b.engine()
b.bike_engine()
```

```python
# Hybrid inheritance
class A:
  def method_A(self):
    print("Method A")
class B(A):
  def method_B(self):
    print("Method B")
class C(A):
  def method_C(self):
    print("Method C")
class D(B, C):
  def method_D(self):
    print("Method D")
    print ("Hybrid inheritance")
d = D()
d.method_A()
d.method_B()
d.method_C()
d.method_D()
```
