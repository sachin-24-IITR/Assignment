```python
# Q1. Create a vehicle class with an init method having instance variables as name_of_vehicle, max_speed and average_of_vehicle.
```

```python
class vehicle:
  def __init__ (self, name_of_vehicle,max_speed,avg_of_vehicle):
    self.name_of_vehicle=name_of_vehicle
    self.max_speed=max_speed
    self.avg_of_vehicle=avg_of_vehicle
v=vehicle("car",200,100)
print(v.name_of_vehicle)
print(v.max_speed)
print(v.avg_of_vehicle)
```

```python
#Q3. What is multiple inheritance? Write a python code to demonstrate multiple inheritance.
```

```python
Multiple inheritance occurs when a class inherits from more than one parent class. In Python, a subclass can inherit attributes and methods from multiple parent classes.
```

```python
class vehicle:
  def __init__(self,name):
    self.name = name
  def brand(self):
    print (f"{self.name} is an EV")
class speed:
  def __init__ (self, max_speed):
     self.max_speed = max_speed
  def maxspeed(self):
    print (f"The maximum speed of {self.name} is {self.max_speed} km/h")
#Multiple inheritance of class vehilce and speed
class car (vehicle,speed):
  def __init__ (self, name, max_speed):
    vehicle.__init__(self, name)
    speed.__init__(self, max_speed)
C= car("Tesla", 300)
C.brand()
C.maxspeed()
```

```python
#Q4. What are getter and setter in python? Create a class and create a getter and a setter method in this class.
```

```python
Getter and Setter are methods used to access and modify the value of variables in a class. These are part of the Encapsulation concept in Object-Oriented Programming (OOP), where we hide the internal state of the object and provide methods to access and modify that state.

Getter: A method that allows you to access the value of a variable.
Setter: A method that allows you to modify the value of a variable.
```

```python
class person:
  def __init__(self, name, age):
    self.__name=name
    self.__age=age
  def get_name(self):
    return self.__name
  def set_name(self, name):
     self.__name=name
  def get_age(self):
    return self.__age
  def set_age(self,age):
    if age>0:
      self.__age=age
    else:
      print ("Age must be positive")
A= person ("Sachin", 27)
print (A.get_name())
print (A.get_age())
print("******")
A.set_name("Kumar")
A.set_age(28)
print (A.get_name())
print (A.get_age())
print ("*******")
A.set_age(-25)
```

```python
#Q5.What is method overriding in python? Write a python code to demonstrate method overriding.
```

```python
Method Overriding is a feature in Object-Oriented Programming (OOP) where a method in a child class has the same name, parameters, and functionality as a method in the parent class, but the child class method provides its own implementation. The child class overrides the method of the parent class.
```

```python
class vehicle:
  def motor(self):
    print ("motor is a vehicle")
class car (vehicle):
  def motor (self):
    print ("car is a motor vehicle")
class bus (vehicle):
  def motor (self):
    print ("bus is a motor vehicle")
v= vehicle()
c= car()
b= bus()
print (v.motor())
print (c.motor())  # car and bus class will override motor method
print (b.motor())
```
